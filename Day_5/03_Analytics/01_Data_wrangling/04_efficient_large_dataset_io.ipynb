```python
import pandas as pd
import io
import numpy as np
csv_data = "id,val\n1,10\n2,20\n3,30\n4,40\n5,50"
chunk_size = 2
total_sum = 0

```

```python
# Simuleted Chunk Reading
for chunk in  pd.read_csv(io.StringIO(csv_data), chunksize=chunk_size):
    total_sum += chunk['val'].sum()
print(f"Cumputed Total via Chunk: {total_sum}")
```
