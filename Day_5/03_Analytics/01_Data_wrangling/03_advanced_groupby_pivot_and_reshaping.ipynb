```python
import pandas as pd
import numpy as np
df_agg = pd.DataFrame({
    'department':['IT','IT','HR','HR','Sales','Sales'],
    'salary':[85000,92000,50000,54000,65000,70000],
    'tenure':[3,5,2,6,4,7]
})
print(df_agg)
```

```python
print("===Mathematical Description===")
print(df_agg.describe())
print("===Information About Data type in this table===")
print(df_agg.info())
```

```python
summary_df = df_agg.groupby('department').agg(
    mean_salary = ('salary','mean'),
    max_tenure = ('tenure','max')
).reset_index()

print(summary_df)
```

```python
df_pivot = pd.DataFrame({
    'device':['Mobile','Desktop','Mobile','Desktop','Mobile','Desktop'],
    'region':['North','North','South','South','North','South'],
    'sales':[200,450,300,500,250,400]
})
print(df_pivot)
```

```python
result = pd.pivot_table(
    df_pivot,
    values="sales",
    index="device",
    columns="region",
    aggfunc="sum",
    margins=True,
)
print(result)
```

```python
df_wide = pd.DataFrame({
    'product':['widget A','widget B'],
    'Jan':[100,150],
    'Feb':[120,170]
})
print(df_wide)
```

```python
result = df_wide.melt(
    id_vars="product",
    value_vars=["Jan","Feb"],
    var_name="Month",
    value_name="Sales"
)
print(result)
```
