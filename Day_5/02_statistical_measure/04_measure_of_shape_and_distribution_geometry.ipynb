## Measure of shape and Distribution Geometry

Central tendency measures the middle

Dispersion measures the spread

Position measures relative rank

meaures of shape describe the structural geometry, asymmetry and tail thickness of a distribution

1. Skewness (Asymmetry of Distribution)

```python
import math
import numpy as np
import scipy.stats as stats
```

```python
salaries = [30,35,40,45,250]

skew_scipy_pop = stats.skew(salaries, bias=True)
skew_scipy_sample = stats.skew(salaries, bias=False)

print(f"Scipy Python Population(True): {skew_scipy_pop:.4f}")
print(f"Scipy Python Sample(False): {skew_scipy_sample:.4f}")
```
