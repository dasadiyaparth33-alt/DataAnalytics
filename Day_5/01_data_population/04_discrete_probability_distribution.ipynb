# Discrete Probability Distributions

A discrete random variable $X$ is a random variable whose possible values either constitute a finite set or a countably infinite sequence (e.g., counts $0, 1, 2, \dots$).

The behavior of $X$ is completely defined by its **Probability Mass Function (PMF)**, denoted as $P(X = x)$ or $p(x)$:

$$P(X = x) \ge 0 \quad \text{and} \quad \sum_{x} P(X = x) = 1$$

---

## 1. The Anatomy of a Bernoulli Trial

A Bernoulli trial is an experiment with exactly two possible outcomes. Let's define our random variable $X$:

- **Success** (which we assign the value $x = 1$) happens with probability $p$.
- **Failure** (which we assign the value $x = 0$) happens with the remaining probability, which is $1 - p$ (often denoted as $q$).

---

## 2. Key Parameters & Conditions

- $x \in \{0, 1\}$
- $0 \le p \le 1$
- $q = 1 - p$

---

## 3. The Probability Mass Function (PMF)

Because there are only two outcomes, writing out the PMF is super clean. It just says:

$$
P(X = x) = 
\begin{cases} 
p & \text{if } x = 1 \\ 
1 - p & \text{if } x = 0 
\end{cases}
$$

Alternatively, it can be written in a single compact equation:

$$P(X = x) = p^x (1 - p)^{1 - x} \quad \text{for } x \in \{0, 1\}$$

## Bernoulli Distribution

The Bernoulli distribution is a discrete probability distribution that models a single experiment (called a Bernoulli trial) which has only two possible outcomes:

• 1 (Success): occurs with probability p

• 0 (Failure): occurs with probability q = 1 - p (where 0 ≤ p ≤ 1)


## PMF Formula

The PMF $P(X = x)$ or $f(x; p)$ is defined as:


Alternatively, it can be written in a single compact equation:

$$P(X=x) = p^{x}(1-p)^{1-x} \quad \text{for } x \in \{0, 1\}$$

---

### Key Parameters & Variables

- **$x$**: The random variable representing the outcome, where $x = 1$ is success and $x = 0$ is failure.
- **$p$**: The probability of getting a success, where $0 \le p \le 1$.
- **$1 - p$** (often denoted as $q$): The probability of getting a failure.

```python
p_bernouli = 0.65

k_bern = np.array([0,1])

pmf_bern = stats.bernoulli.pmf(k_bern, p=p_bernouli)

print("1. Bernouli Distribution")
print(f"Success Parameter: {p_bernouli:.2f}")
print(f"P(X=0) [Failur]: {pmf_bern[0]:.4f}")
print(f"P(X=1) [Success]: {pmf_bern[1]:.4f}")
print(f"Theoritical Mean: {stats.bernoulli.mean(p=p_bernouli):.4f}")
print(f"Theoritical Var: {stats.bernoulli.var(p=p_bernouli):.4f}")

```

### The Binomial Formula

To find the probability of getting exactly $k$ successes in $n$ independent trials, you use the *Probability Mass Function (PMF)*:

$$P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}$$

```python
from scipy.stats import binom

n = 5
p = 0.7

prob_4 = binom.pmf(k=4, n=n, p=p)
print(f"Exact Probability of making 4 shots: {prob_4:.4f}(~{prob_4 * 100:.1f}%)")
```

```python
simulations = binom.rvs(n=n, p=p, size=1000)

average_makes = sum(simulations) / len(simulations)
expected_val = n * p
print(f"Average makes: {average_makes:.2f}")
print(f"Expected Value: {expected_val:.2f}")
```

## Poisson Distribution PMF Formula

The Probability Mass Function (PMF) $P(X = k)$ or $f(k; \lambda)$ gives the probability of observing exactly $k$ events in a fixed interval of time or space:

$$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!} \quad \text{for } k \in \{0, 1, 2, 3, \dots\}$$

---

### Key Parameters & Variables

- **$k$**: The actual number of occurrences or successes observed ($k = 0, 1, 2, \dots$). Must be a non-negative integer.
- **$\lambda$ (lambda)**: The average rate (mean number of occurrences) in the specified interval, where $\lambda > 0$.
- **$e$**: Euler's number (the base of natural logarithms), approximately equal to $2.71828$.
- **$k!$**: The factorial of $k$ ($k! = k \times (k-1) \times \dots \times 1$, with $0! = 1$).

---

### Key Statistical Properties

- **Mean / Expected Value:** $E(X) = \lambda$
- **Variance:** $\operatorname{Var}(X) = \lambda$
- **Standard Deviation:** $\sigma = \sqrt{\lambda}$

```python
from scipy.stats import poisson
```

```python
lambda_rate = 3
prob_2 = poisson.pmf(k=2, mu = lambda_rate)
print(f"Probability of getting exactly 2 orders: {prob_2:.4f}(~{prob_2 * 100:.4f}%)")
simulated_hours = poisson.rvs(mu=lambda_rate, size = 10)
print(f"Simulated orders for 10 hours(random value of x): {simulated_hours}")
```

## Geometric Distribution PMF Formula

The Geometric distribution models the number of Bernoulli trials needed to get the **first success**. 

There are two common formulations depending on how the random variable is defined:

### 1. Trials until First Success ($k \in \{1, 2, 3, \dots\}$)
Here, $X$ represents the **trial number of the first success** (e.g., success happens on the $k$-th trial after $k - 1$ failures):

$$P(X = k) = (1 - p)^{k - 1} p \quad \text{for } k \in \{1, 2, 3, \dots\}$$

### 2. Failures before First Success ($y \in \{0, 1, 2, \dots\}$)
Here, $Y$ represents the **number of failures before the first success occurs**:

$$P(Y = y) = (1 - p)^{y} p \quad \text{for } y \in \{0, 1, 2, \dots\}$$

---

### Key Parameters & Variables

- **$p$**: Probability of success on each independent trial, where $0 < p \le 1$.
- **$1 - p$** (or $q$): Probability of failure on each trial.
- **$k$**: Total number of trials up to and including the first success.
- **$y$**: Number of failures preceding the first success ($y = k - 1$).

---

### Key Statistical Properties

| Property | Trials until First Success ($X$) | Failures before First Success ($Y$) |
| :--- | :--- | :--- |
| **Mean / $E(\cdot)$** | $\frac{1}{p}$ | $\frac{1 - p}{p}$ |
| **Variance / $\operatorname{Var}(\cdot)$** | $\frac{1 - p}{p^2}$ | $\frac{1 - p}{p^2}$ |
| **CDF / $P(\cdot \le t)$** | $1 - (1 - p)^k$ | $1 - (1 - p)^{y + 1}$ |

---

### Notable Feature: Memoryless Property

The geometric distribution is the only discrete distribution that possesses the **memoryless property**:

$$P(X > s + t \mid X > s) = P(X > t)$$
